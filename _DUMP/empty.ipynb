```python
import matplotlib.pyplot as plt

# Read data from file
generation = []
cost = []
with open('generation_cost.txt', 'r') as file:
    for line in file:
        gen, c = line.strip().split(',')
        generation.append(int(gen))
        cost.append(float(c.strip('()')))

# Plot the data
plt.plot(generation, cost, marker='o', linestyle='-')
plt.xlabel('Generation')
plt.ylabel('Best Cost')
plt.title('Best Cost vs Generation')
plt.grid(True)
plt.show()
```
